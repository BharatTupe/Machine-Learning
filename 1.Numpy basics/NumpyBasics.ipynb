```python
import time 
import numpy as np

size = 100000

# add elements of array a and b
def addition_using_list():
    t1 = time.time()
    a = range(size)
    b = range(size)
    c = [a[i]+b[i] for i in range(size)]
    t2 = time.time()
    return (t2-t1)
def addition_using_numpy():
    t1 = time.time()
    a = np.arange(size)
    b = np.arange(size)
    c = a + b
    t2 = time.time()
    return (t2-t1)
t_list = addition_using_list()
t_numpy = addition_using_numpy()
print("List=", t_list * 1000)  ##milliseconds
print("numpy=",t_numpy *1000) ##milliseconds
```
