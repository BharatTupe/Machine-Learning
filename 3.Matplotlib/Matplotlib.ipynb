# Line Chart

# Bar chart

```python
plt.bar(x=["red","blue","green"],height=[20,40,30],color="orange",edgecolor="blue")
plt.title("this is the bargraph")
plt.xlabel("colors")
plt.ylabel("Intensity")
plt.show()
```

```python
categories = ['Mon', 'Tue', 'Wed', 'Thu', 'Fri']
sales      = [42, 75, 58, 90, 67]

plt.bar(categories, sales)
plt.show()
```

```python
bars = plt.bar(categories, sales,
        color='steelblue',
        width=0.60)
```

```python
# Chart titles & axis labels
plt.title('Weekly Sales', fontsize=14, fontweight='bold')
plt.xlabel('Day', fontsize=11)
plt.ylabel('Units Sold', fontsize=11)

# Annotate each bar with its value
bars = plt.bar(categories, sales)
for bar in bars:
    h = bar.get_height()
    plt.text(bar.get_x() + bar.get_width()/2,
            h + 1, str(h),
            ha='center', va='bottom', fontsize=9)

plt.tight_layout()
plt.show()
```

```python
categories = ['Q1', 'Q2', 'Q3', 'Q4']
product_a  = [30, 55, 48, 70]
product_b  = [45, 38, 62, 55]

x     = np.arange(len(categories))
width = 0.35  # width of each bar

fig, ax = plt.subplots()
ax.bar(x - width/2, product_a, width, label='Product A')
ax.bar(x + width/2, product_b, width, label='Product B')

ax.set_xticks(x)
ax.set_xticklabels(categories)
ax.legend()
plt.show()
```

```python
bottom1 = product_a
bottom2 = [a+b for a,b in zip(product_a, product_b)]
product_c = [40,30,60,50]
plt.bar(x, product_a, label='Segment A')
plt.bar(x, product_b, bottom=bottom1, label='Segment B')
plt.bar(x, product_c, bottom=bottom2, label='Segment C')
plt.legend()
```

```python
# 100% stacked — normalise to percentages first
totals = [a+b+c for a,b,c in zip(product_a, product_b, product_c)]
pct_a  = [v/t*100 for v,t in zip(product_a, totals)]
pct_b  = [v/t*100 for v,t in zip(product_b, totals)]
pct_c = [v/t*100 for v,t in zip(product_c, totals)]

# Bottom for C
bottom_c = [a+b for a,b in zip(pct_a, pct_b)]

plt.bar(x, pct_a, label='A')
plt.bar(x, pct_b, bottom=pct_a, label='B')
plt.bar(x,pct_c,bottom=bottom_c,label="C")
plt.ylabel('%')
```

```python
# plt.barh(y, width) — note the argument order!
items  = ['Python', 'JavaScript', 'Rust', 'Go', 'Java']
scores = [92, 78, 65, 71, 60]

plt.barh(items, scores, color='steelblue')
plt.xlabel('Popularity score')
plt.title('Language Rankings')

# Sort trick: zip, sort, unzip
paired = sorted(zip(scores, items))
scores, items = zip(*paired)
plt.barh(items, scores)
plt.show()
```

```python
# plt.barh(y, width) — note the argument order!
items  = ['Python', 'JavaScript', 'Rust', 'Go', 'Java']
scores = [92, 78, 65, 71, 60]

plt.barh(items, scores, color='steelblue')
plt.xlabel('Popularity score')
plt.title('Language Rankings')

# Sort trick: zip, sort, unzip
paired = sorted(zip(scores, items))
scores, items = zip(*paired)
plt.barh(items, scores)
plt.show()
```

```python
items  = ['Python', 'JavaScript', 'Rust', 'Go', 'Java']
scores = [92, 78, 65, 71, 60]

# Combine + sort ascending
paired = sorted(zip(scores, items))

# Separate again
scores, items = zip(*paired)

# Plot
plt.barh(items, scores, color='steelblue')

plt.xlabel('Popularity score')
plt.title('Ascending Order')

plt.show()
```

```python
items  = ['Python', 'JavaScript', 'Rust', 'Go', 'Java']
scores = [92, 78, 65, 71, 60]

# Sort descending
paired = sorted(zip(scores, items), reverse=True)

# Unzip
scores, items = zip(*paired)

# Plot
plt.barh(items, scores, color='orange')

plt.xlabel('Popularity score')
plt.title('Descending Order')
```

```python
data = np.random.normal(50, 10, 1000)

plt.hist(
    data,
    bins=20,
    color='orange',
    edgecolor='black',
    linewidth=2,
    alpha=0.7,
    density=True,
    histtype='bar',
    rwidth=0.9
)

plt.title('Histogram Example')
plt.xlabel('Values')
plt.ylabel('Frequency')

plt.grid(True)

plt.show()
```

```python
from scipy.stats import gaussian_kde

fig, ax = plt.subplots()
ax.hist(data, bins=25, density=True,
        alpha=0.7, color='steelblue')

# KDE smooth curve overlay
kde = gaussian_kde(data)
x   = np.linspace(data.min(), data.max(), 200)
ax.plot(x, kde(x), color='red', lw=2)
```

```python
plt.hist(data, bins=30,
        cumulative=True,
        density=False)
plt.ylabel('Cumulative count')
```

# BoxPlot

```python
data = np.random.normal(50, 15, 200)
plt.boxplot(data,notch=True,vert=True)
plt.show()
```

```python
data = np.random.normal(50, 10, 200)

plt.boxplot(
    data,
    patch_artist=True,
    notch=True,
    widths=0.5,
    showmeans=True,

    boxprops=dict(
        facecolor='orange',
        color='black',
        linewidth=2
    ),

    medianprops=dict(
        color='red',
        linewidth=3
    ),

    whiskerprops=dict(
        color='green',
        linestyle='--',
        linewidth=2
    ),

    capprops=dict(
        color='blue',
        linewidth=2
    ),

    flierprops=dict(
        marker='o',
        markerfacecolor='purple',
        markersize=8
    ),

    meanprops=dict(
        marker='D',
        markerfacecolor='black',
        markersize=10
    )
)

plt.title('Customized Box Plot')

plt.show()
```

# Violen Plot

```python
data = np.random.normal(50, 10, 200)

plt.violinplot(data,showmeans=True,showmedians=True,quantiles=[[0.25, 0.5, 0.75]])
plt.title("Basic Violin Plot")
plt.show()
```

# Piechart

```python
data = [30, 20, 40, 10]

plt.pie(data)

plt.show()
```

```python
data = [35, 25, 20, 20]

labels = ['Python', 'Java', 'C++', 'Go']

colors = ['orange', 'blue', 'green', 'red']

explode = [0, 0.1, 0, 0]

plt.pie(
    data,
    labels=labels,
    colors=colors,
    explode=explode,
    autopct='%1.1f%%',
    shadow=True,
    startangle=90,

    wedgeprops={
        'edgecolor': 'black',
        'linewidth': 2
    },

    textprops={
        'fontsize': 12,
        'color': 'black'
    }
)

plt.title('Programming Language Popularity')

plt.show()
```

```python
data = [40,30,20,10]

plt.pie(data)

centre_circle = plt.Circle((0,0),0.70,color='white')

fig = plt.gcf()

fig.gca().add_artist(centre_circle)

plt.show()
```

```python
sales = [450, 300, 150, 100]

labels = ['Electronics', 'Clothing', 'Food', 'Books']

explode = [0.1, 0, 0, 0]

colors = ['gold', 'lightblue', 'lightgreen', 'orange']

plt.pie(
    sales,
    labels=labels,
    colors=colors,
    explode=explode,
    autopct='%1.1f%%',
    shadow=True,
    startangle=140,
    counterclock=False,

    wedgeprops={
        'edgecolor': 'black',
        'linewidth': 2
    },

    textprops={
        'fontsize': 12
    }
)

plt.title('Sales Distribution')

plt.show()
```

```python
months = ['Jan','Feb','Mar','Apr','May']

sales = [100,120,90,140,170]

plt.plot(
    months,
    sales,
    color='blue',
    linewidth=3
)

plt.fill_between(
    months,
    sales,
    color='lightblue',
    alpha=0.6
)

plt.title("Monthly Sales")
plt.xlabel("Months")
plt.ylabel("Sales")

plt.show()
```
