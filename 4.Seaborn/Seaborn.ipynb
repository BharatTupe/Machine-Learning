```python
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import pandas as pd
```

```python
df = sns.load_dataset("tips")
```

```python
df.head()
```

# Line plot

```python
sns.lineplot(x="size",y="tip",data=df)
plt.title("Line Plot")
plt.xlabel("size")
plt.ylabel("tip")
plt.show()
```

# Scatterplot

```python
sns.scatterplot(x="tip",y="total_bill",hue="sex",data=df,size="size",alpha=0.7,style="time")
plt.title("Scatter Plot")
plt.xlabel("tip")
plt.ylabel("total_bill")
plt.show()
```

# Point Plot

```python
sns.pointplot(x="size",y="tip",data=df)
plt.show()
```

# Count Plot

```python
sns.countplot(x="sex",color="lightblue",data=df)
plt.show()
```

# Bar Plot

```python
sns.barplot(x="total_bill",y="sex",data=df,estimator="sum")
```

```python
sns.barplot(x="day",y="total_bill",data=df,estimator="sum",hue="sex")
```

# Boxplot

```python
sns.boxplot(y="tip",data=df)
```

```python
sns.boxplot(x="tip",data=df)
```

```python
sns.boxplot(x="sex",y="tip",data=df)
```

```python
sns.boxplot(x="day", y="total_bill", hue="sex", data=df)
```

```python
sns.boxplot(
    data=df, x="sex", y="total_bill",
    notch=True, showcaps=False,
    flierprops={"marker": "x"},
    boxprops={"facecolor": (.3, .5, .7, .5)},
    medianprops={"color": "r", "linewidth": 2},
)
```

# Violen Plot

```python
sns.violinplot(data=df, x="sex", y="total_bill")
```

```python
sns.violinplot(data=df, x="sex", y="total_bill",inner="quart")
```

# Histogram

```python
sns.histplot(data=df, x="total_bill",bins=5)
```

```python
sns.histplot(data=df, x="total_bill",bins=5,binwidth=3)
```

```python
sns.histplot(data=df, x="total_bill",bins=5,kde=True)
```

```python
sns.histplot(data=df, x="total_bill", hue="sex", multiple="stack")
```

```python
sns.histplot(data=df, x="total_bill", hue="sex", multiple="stack",element="step")
```

```python
sns.histplot(data=df, x="total_bill", hue="sex", multiple="stack",element="poly")
```
